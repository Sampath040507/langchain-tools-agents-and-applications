```python
#we are ready with tool and llm, now we can create an agent out of them:
from langchain_classic.agents import create_react_agent, AgentExecutor
from langchain_classic import hub
#1) hub - To retrieve reusable prompts from the LangChain Hub.
#eg:
# prompt = hub.pull("hwchase17/react")
# This retrieves a ReAct prompt template that teaches the LLM how to reason through a task and use tools.

#2) create_react_agent - To create a ReAct agent by connecting your LLM, tools, and prompt.
# agent = create_react_agent(llm, tools, prompt):
# It creates the agent that decides which tool to use and what action to take based on the prompt and the LLM's output.

#3)AgentExecutor - To execute and manage the agent's tool-use loop.
# agent_executor = AgentExecutor(
#     agent=agent,
#     tools=tools,
#     verbose=True
# )
# It runs the agent, executes requested tools, passes their results back to the agent, and continues until the agent finishes or reaches a stopping limit.

# Remember: hub gets the instructions, create_react_agent builds the agent, and AgentExecutor runs it.
```

```python
# Step 2: Pull the ReAct prompt from LangChain Hub

# prompt = hub.pull(
#     "hwchase17/react",
#     dangerously_pull_public_prompt=True
# )

 # pulls the standard ReAct agent prompt

# i.e prompt's name which we are pulling is react but why react needs to be pulled since the agent what we are going to create is a react agent: reAct (reasoning + action)
#so many ai agents exist and behind every ai agent there exists a design pattern whose name here is react and the user who designed the prompt is harrison chase
#harrison chase is the creator of langchain and his account is hwchase17 and react- he wrote a predefined prompt for react agents. so using this we build our agent

#The error says that pulling a public prompt by its owner/name is disabled by default because the prompt could contain untrusted serialized LangChain objects.
#to overcome we use dangerously_pull_public_prompt but The new error reveals that hub.pull() in your installed version does not accept dangerously_pull_public_prompt.
#My previous suggestion to pass that argument to hub.pull() was incorrect.
#since Your output also says hub.pull() is deprecated. The newer approach is to use Client from langsmith.
```

```python
from langsmith import Client

client = Client()

prompt = client.pull_prompt(
    "hwchase17/react",
    dangerously_pull_public_prompt=True
)

#LangSmith is a platform developed by the LangChain team for debugging, monitoring, testing, and evaluating LLM applications and AI agents.
# Think of it this way:
# - LangChain helps you build an AI application or agent.
# - LangSmith helps you understand, debug, and evaluate how that application behaves.

# Main purposes of LangSmith
# Purpose	What it means
# Tracing	Tracks the steps an LLM application performs, including model calls and tool executions.
# Debugging	Helps identify where an application fails and inspect inputs and outputs.
# Evaluation	Measures how well your application answers questions or completes tasks.
# Monitoring	Helps observe application performance and behavior in production.
# Prompt management	Lets you store, version, share, and retrieve prompts.

# The Client object lets your Python code interact with LangSmith services, including retrieving prompts. Public prompt retrieval may require additional security configuration, as you encountered.

# Remember: LangChain builds the agent; LangSmith helps you inspect, test, and improve it. LangSmith is a separate platform, not an LLM or an agent itself.
```

```python
# Step 3: Create the ReAct agent manually with the pulled prompt
agent = create_react_agent(
    llm=llm,
    tools=[search_tool],
    prompt=prompt,
    stop_sequence=False  #this is mandatory to write otherwise while running the agentexecutor, i mean druing its invoke: An error will occur because your
    #OpenAI model does not support the stop parameter that your current ReAct agent implementation is sending to it.
)
 #there are many agents but we chose to go with react design pattern and which has gpt as llm with a search tool

#here tools are specified adn in agentexecutor class also mentioned why?
# here:
# Purpose: To create the agent and tell it which tools it is allowed to consider using.
# The agent uses:
# - llm — to decide what action to take.
# - tools — to understand which tools are available and their descriptions.
# - prompt — to follow the ReAct reasoning-and-action format.
# For example, the LLM might decide: "I need to search the web to answer this question."
# The agent can identify search_tool as the appropriate tool, but this function alone does not execute the entire agent workflow.
```

```python
# Step 4: Wrap it with AgentExecutor
agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tool],
    verbose=True,
    handle_parsing_errors=True
)
#so agent thinks and plans whereas agentexecutor executes those planned things like say hitting the apis to fectch info or executing or using tools to perform certain tasks and
#returns the results back to agent

#here specifying the tools too:
# The executor needs the actual tool implementation so it can execute the agent's requested action.
# For example, it:
# 1. Gets the agent's proposed action.
# 2. Executes search_tool when requested.
# 3. Passes the tool result back to the agent.
# 4. Repeats until the agent finishes or reaches a stopping limit.

# create_react_agent() - Configures the agent with the LLM, prompt and available tools.
# AgentExecutor() - Runs the agent and executes its requested tools.
# finaly,the agent knows what tools it can request, while the executor manages their execution.
# The same tool is supplied to both because the agent needs to know about it, and the executor needs access to execute it.
```

```python
#verbose=True displays additional execution details in your notebook or console, which is useful for learning and debugging.

# Step 5: Invoke
response = agent_executor.invoke({"input": "3 ways to reach goa"})
print(response)  
```
